```python
import torch
import torch.nn as nn
```

starting with an example of the neural network having\
5 input features and 10 corresponding rows(instance)

```python
class Model(nn.Module):
    def __init__(self, num_features):

        super().__init__()
        self.linear=nn.Linear(num_features, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        out=self.linear(features)
        out=self.sigmoid(out)

        return out

```

```python
#create the dataset
features=torch.rand(10,5)

## create the Model 
model=Model(features.shape[1])

## call mode for forward pass
#model.forward(features) also work here pytorch uses the magic method __call__
model(features) ## standard and required way to call
```

```python
# show model weights and bias
print(model.linear.weight)
print(model.linear.bias)
```

```python
from torchinfo import summary
```

```python
summary(model, input_size=(10,5))
```

```python
class Model2(nn.Module):
    def __init__(self, num_features):

        super().__init__()
        self.linear=nn.Linear(num_features, 3)
        self.relu=nn.ReLU()
        self.linear2=nn.Linear(3,1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        out=self.linear(features)
        out=self.relu(out)
        out=self.linear2(out)
        out=self.sigmoid(out)

        return out
```

```python
#create the dataset
features=torch.rand(10,5)

## create the Model 
model2=Model2(features.shape[1])

## call mode for forward pass

model2(features) ## standard and required way to call
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv")
```

```python
df.head()
```

```python
df.drop(columns=['id','Unnamed: 32'],inplace=True)
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:], df.iloc[:,0], test_size=0.2)
```

```python
scaler=StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

encoder=LabelEncoder()
y_train=encoder.fit_transform(y_train)
y_test=encoder.transform(y_test)
```

```python
type(X_train[0][9])
```

```python
X_train_tensor=torch.from_numpy(X_train).to(torch.float32)
X_test_tensor=torch.from_numpy(X_test).to(torch.float32)
y_train_tensor=torch.from_numpy(y_train).to(torch.float32)
y_test_tensor=torch.from_numpy(y_test).to(torch.float32)
```

```python
class mySimpleNN(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        self.linear = nn.Linear(num_features,1)
        self.sigmoid=nn.Sigmoid()

    def forward(self, features):
        out=self.linear(features)
        out=self.sigmoid(out)
        return out

    def loss_function(self, y, y_pred):
        # Clamp predictions to avoid log(0)
        epsilon=1e-7
        y_pred=torch.clamp(y_pred, epsilon, 1-epsilon)
        #calculate the loss
        loss = -(y*torch.log(y_pred) + (1-y)*(1-y_pred)).mean()
        return loss
```

important paramerters

```python
learning_rate = 0.1
epochs = 25
```

```python
# create the model 
model=mySimpleNN(X_train_tensor.shape[1])

# define loop 
for epoch in range(epochs):

    # forward pass
    y_pred= model(X_train_tensor)
    #print(y_pred)
    
    # loss calculate
    loss = model.loss_function(y_train_tensor, y_pred)

    # backward pass
    loss.backward()

    # parameters update
    with torch.no_grad():
        model.linear.weight -= learning_rate * model.linear.weight.grad
        model.linear.bias -= learning_rate*model.linear.bias.grad

    ## zero gradients
    model.linear.weight.grad.zero_()
    model.linear.bias.grad.zero_()

    # print the loss in each epoch
    print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

Now we are making our model fully automatic in calculating the loss function and parameter optimization step... by using the inbuilt function...

```python
## defining the loss function
loss_function = nn.BCELoss()


class mySimpleNN2(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        self.linear = nn.Linear(num_features,1)
        self.sigmoid=nn.Sigmoid()

    def forward(self, features):
        out=self.linear(features)
        out=self.sigmoid(out)
        return out
    
# create the model 
model=mySimpleNN2(X_train_tensor.shape[1])

# defining the optimizer
optimizer=torch.optim.SGD(model.parameters(), lr=learning_rate)

for epoch in range(epochs):

    # forward pass
    y_pred= model(X_train_tensor)
    #print(y_pred)
    
    # loss calculate
    loss = loss_function( y_pred, y_train_tensor.view(-1,1))

    ## zero gradients
    optimizer.zero_grad()

    # backward pass
    loss.backward()

    # parameters update
    optimizer.step()

    # print the loss in each epoch
    print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```
